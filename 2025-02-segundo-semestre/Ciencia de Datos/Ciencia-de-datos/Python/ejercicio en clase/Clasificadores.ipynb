```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import warnings
```

```python
warnings.filterwarnings("ignore")
plt.style.use("fivethirtyeight")
sns.set_style("darkgrid")
```

```python
from sklearn.linear_model import LinearRegression
import xgboost
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
dfBikes = pd.read_csv(
    "http://kanaries-app.s3.ap-northeast-1.amazonaws.com/public-datasets/bike_sharing_dc.csv",
    parse_dates=["date"],
)
dfBikes
```

```python
dfBikes.dtypes
```

```python
estacion = dfBikes.groupby("season").size().index
estacion
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
codificador = LabelEncoder()
dfBikes["season"] = codificador.fit_transform(dfBikes["season"])
dfBikes["holiday"] = codificador.fit_transform(dfBikes["holiday"])
dfBikes["am or pm"] = codificador.fit_transform(dfBikes["am or pm"])
dfBikes
```

```python
train, test = train_test_split(
    dfBikes, test_size=0.4, stratify=dfBikes["season"], random_state=10
)
```

```python
# Agrupamiento por año, mes y periodo del día (am/pm)
train_year_month_ampm = train.groupby(['year', 'month', 'am or pm'])['count'].sum()
train_year_month_ampm
```

```python
# Crear DataFrame para mejor visualización
df_agrupado = train_year_month_ampm.reset_index()
df_agrupado['periodo'] = df_agrupado['am or pm'].map({0: 'AM', 1: 'PM'})
df_agrupado
```

```python
# Gráfica de barras agrupadas por año, mes y periodo del día
plt.figure(figsize=(15, 8))

# Crear pivot table para mejor visualización
pivot_data = df_agrupado.pivot_table(values='count', index=['year', 'month'], columns='periodo', aggfunc='sum')

# Gráfica de barras
pivot_data.plot(kind='bar', figsize=(15, 8), color=['lightblue', 'orange'])
plt.title('Uso de Bicicletas por Año, Mes y Periodo del Día (AM/PM)', fontsize=16)
plt.xlabel('Año - Mes', fontsize=12)
plt.ylabel('Cantidad de Bicicletas', fontsize=12)
plt.legend(title='Periodo del Día')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Gráfica alternativa con seaborn - Heatmap
plt.figure(figsize=(12, 6))

# Crear pivot para heatmap
heatmap_data = df_agrupado.pivot_table(values='count', index='month', columns=['year', 'periodo'], aggfunc='sum')

sns.heatmap(heatmap_data, annot=True, fmt='d', cmap='YlOrRd', cbar_kws={'label': 'Cantidad de Bicicletas'})
plt.title('Mapa de Calor: Uso de Bicicletas por Mes, Año y Periodo (AM/PM)', fontsize=14)
plt.xlabel('Año - Periodo del Día', fontsize=12)
plt.ylabel('Mes', fontsize=12)
plt.tight_layout()
plt.show()
```
