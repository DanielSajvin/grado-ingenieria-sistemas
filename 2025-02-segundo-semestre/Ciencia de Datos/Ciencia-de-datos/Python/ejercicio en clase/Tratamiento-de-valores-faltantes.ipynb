```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
datos = sns.load_dataset("titanic")
datos
```

```python
datos.info()
```

```python
datos.columns
```

```python
datos.drop("embarked", axis=1, inplace=True)
datos.drop("class", axis=1, inplace=True)
datos.drop("who", axis=1, inplace=True)
datos.drop("adult_male", axis=1, inplace=True)
datos.drop("deck", axis=1, inplace=True)
datos.drop("embark_town", axis=1, inplace=True)
datos.drop("alive", axis=1, inplace=True)
datos.drop("alone", axis=1, inplace=True)
datos
```

```python
from sklearn.preprocessing import LabelEncoder

codificador = LabelEncoder()

datos['sex'] = codificador.fit_transform(datos['sex'])
datos
```

```python
copiaDatos = datos.copy()
```

```python
# Dividir el conjunto en X (explicativas) y Y (repuesta)
y = datos["survived"]
datos.drop("survived", axis=1, inplace=True)
datos
```

```python
datos.info()
```

```python
# recuento de cuantos valores nulos hay
print(datos.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(datos, y, test_size=0.3)
```

```python
from sklearn.linear_model import LogisticRegression

# crear modelo 
lr = LogisticRegression()
```

```python
lr.fit(x_train, y_train)
```

```python
# eliminar columnas con valores faltantes 
datosActualizados = datos.dropna(axis=1)
datosActualizados
```

```python
x_train, x_test, y_train, y_test = train_test_split(datosActualizados, y, test_size=0.3)
lr.fit(x_train, y_train)

# calcular la prediccion 
pred = lr.predict(x_test)
```

```python
from sklearn import metrics
```

```python
print(metrics.accuracy_score(pred, y_test))
```

```python
# Imputar con media(promedio de los datos), 
# mediana(valor que se encuenetra jsutamente en la posición central de la serie de datos ordenada), 
# moda(el mas frecuente, puede ser unimodal, bimodal, multimodal o no haber moda)
```

```python
datosRelleno = copiaDatos.copy()
datosRelleno['age'] = datosRelleno['age'].fillna(datosRelleno['age'].mean())
datosRelleno.info()
```

```python
from sklearn import metrics
```

```python
# segmentar los grupos en entrenamiento y pruebas
y = datosRelleno["survived"]
datosRelleno.drop("survived", axis=1, inplace=True)

x_train, x_test, y_train, y_test = train_test_split(datosRelleno, y, test_size=0.3)

lr.fit(x_train, y_train)

pred = lr.predict(x_test)
print(metrics.accuracy_score(pred, y_test))
```

```python
# Otras formars de imputacion
# interpolacion linal
# rellevar con el dato de la siguiente fila
# agregar una columna
```

```python
imputados = copiaDatos.copy()
imputados['ageMissing'] = imputados['age'].isnull()
imputados.info()
```

```python
from sklearn.impute import SimpleImputer
```

```python
y = imputados['survived']
imputados.drop('survived', axis=1, inplace=True)
```

```python
imputador = SimpleImputer(strategy='median')
nuevosDatos = imputador.fit_transform(imputados)
imputados.info()
```

```python
nuevosDatos
```

```python
columnas = imputados.columns
imputados = pd.DataFrame(nuevosDatos, columns=columnas)
imputados
```

```python
imputados.info()
```

```python
x_train, x_test, y_train, y_test = train_test_split(imputados, y, test_size=0.3)

lr.fit(x_train, y_train)

pred = lr.predict(x_test)

print(metrics.accuracy_score(pred, y_test))
```

```python
copiaDatos.describe()
```

```python
imputados['ageMissing'] = imputados['ageMissing'].astype('bool')
imputados.info()
```

```python
x_train, x_test, y_train, y_test = train_test_split(imputados, y, test_size=0.3)

lr.fit(x_train, y_train)

pred = lr.predict(x_test)

print(metrics.accuracy_score(pred, y_test))
```
