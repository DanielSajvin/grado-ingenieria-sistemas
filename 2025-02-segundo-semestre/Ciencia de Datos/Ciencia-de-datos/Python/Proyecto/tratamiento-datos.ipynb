**edad_al_resultado**, es la columna que requiere mas tratamiento
Las edades son texto. Entonces se tienen que convertir a un solo número para esto mejor se pasará todo a días.
También recordar que tiene 8 valores nulos (non-null 78248). Para los valores nulos se van a rellenar con "0 days" (asumiendo que son recién nacidos o desconocidos) esto por comodidad y para para que la función de conversión no falle.

```python
# Tratamiento de edad_al_resultado

# 1. Se rellenan los 8 valores nulos
df["edad_al_resultado"] = df["edad_al_resultado"].fillna("0 days")
```

```python
# 2. Se crea la función de conversión, es decir, de texto a días
def convertir_edad_a_dias(texto_edad):
    # texto_edad a veces puede ser un float, para evitar errores entonces se convierte a string
    texto_edad = str(texto_edad)

    if "year" in texto_edad or "years" in texto_edad:
        return int(texto_edad.split()[0]) * 365
    elif "month" in texto_edad or "months" in texto_edad:
        return int(texto_edad.split()[0]) * 30
    elif "week" in texto_edad or "weeks" in texto_edad:
        return int(texto_edad.split()[0]) * 7
    elif "day" in texto_edad or "days" in texto_edad:
        return int(texto_edad.split()[0])
    else:
        return 0  # esto porque puede ser 'Unknown' o formatos no reconocidos
```

```python
# 3. Se llama a la función para crear la nueva columna 'edad_en_dias'
df["edad_en_dias"] = df["edad_al_resultado"].apply(convertir_edad_a_dias)
```

```python
# 4. Verificar el resultado (mostramos la columna vieja y la nueva)
print("\n--- Verificación de 'edad_en_dias' ---")
print(
    df[["edad_al_resultado", "edad_en_dias"]].head(10)
)  # mostrar las primeras 10 filas
```

Ahora se pasa al tratamiento de **sexo_al_resultado (sex_upon_outcome)**
Esta columna contiene dos datos: el sexo (Macho/Hembra) y si está esterilizado (Sí/No). También recordar que tiene 2 valores nulos. La intención es simplificar esta columna, para un mejor tratamiento de los datos, eso se realizará de la siguiente manera: 

```python
# 1. Primero se tiene que ver si existen y cuáles son los valores únicos
print("--- Valores Originales de 'sexo_al_resultado' ---")
print(df["sexo_al_resultado"].unique())
```

```python
# 2. Ahora se crea una función para extraer si está esterilizado/a
def es_esterilizado(texto):
    texto = str(texto)  # Convertimos a string para manejar los 'nan'
    if "Neutered" in texto or "Spayed" in texto:
        return "Si"
    elif "Intact" in texto:
        return "No"
    else:
        return "Desconocido"  # Esto es para 'Unknown' y los 'nan'
```

```python
# 3. Función para extraer el sexo
def obtener_sexo(texto):
    texto = str(texto)  # Para evitar errores se convierte a string
    if "Male" in texto:
        return "Macho"
    elif "Female" in texto:
        return "Hembra"
    else:
        return "Desconocido"  # Esto es para 'Unknown' y los 'nan'
```

```python
# 4. Ahora se divide "sexo_al_resultado" en las nuevas columnas aplicando las funciones
df["esterilizado"] = df["sexo_al_resultado"].apply(es_esterilizado)
df["sexo"] = df["sexo_al_resultado"].apply(obtener_sexo)
```

```python
# 5. Se verifica el resultado
print("\n--- Verificación de 'esterilizado' y 'sexo' ---")
print(df[["sexo_al_resultado", "esterilizado", "sexo"]].head(10))
```

Se realizó el tratamiento para las columnas que más lo necesitaban, aún así se dejaron varios valores nulos anteriormente, ahora se manejarán estos valores nulos de la siguiente manera: 

```python
# se eliminan las filas con 'tipo_resultado' = nulo
print(f"Filas antes de eliminar nulos: {len(df)}")

# se eliminan las filas donde 'tipo_resultado' es nulo
df = df.dropna(subset=["tipo_resultado"])

print(f"Filas después de eliminar nulos: {len(df)}")
```
